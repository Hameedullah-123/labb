```python
import heapq   # Import heapq for priority queue (min-heap)

# Graph based on your image (node: [(neighbor, cost)])
graph = {
    'S': [('A', 3), ('B', 2), ('C', 7)],  # S connects to A, B, C
    'A': [('D', 8), ('G', 15)],          # A connects to D and G
    'B': [('G', 20)],                    # B connects to G
    'C': [('G', 6)],                     # C connects to G
    'D': [],                             # D has no outgoing edges
    'G': []                              # G is goal node
}

# UCS function
def ucs(start, goal):
    # Priority queue: (cost, current_node, path_taken)
    pq = [(0, start, [start])]

    visited = set()   # Set to keep track of visited nodes

    # Loop until queue is empty
    while pq:
        # Get node with smallest cost
        cost, node, path = heapq.heappop(pq)

        # If already visited, skip it
        if node in visited:
            continue

        # Mark node as visited
        visited.add(node)

        # If goal reached, return path and cost
        if node == goal:
            return path, cost

        # Check all neighbors of current node
        for neighbor, weight in graph[node]:
            # If neighbor not visited
            if neighbor not in visited:
                # Add new path with updated cost to queue
                heapq.heappush(pq, (cost + weight, neighbor, path + [neighbor]))

    return None   # If no path found

# Run UCS from S to G
result = ucs('S', 'G')

# If result found
if result:
    path, cost = result
    print("Path:", " -> ".join(path))   # Print path
    print("Total Cost:", cost)          # Print total cost
else:
    print("No path found")              # If no solution
```
