```python
import heapq

class Node:
    def __init__(self, name, heuristic=0):
        self.name = name
        self.heuristic = heuristic
        self.edges = {}  # dictionary mapping neighbor Node to edge weight

    def add_edge(self, neighbor, weight):
        self.edges[neighbor] = weight

class Graph:
    def __init__(self):
        self.nodes = {}  # dictionary mapping node name to Node object

    def add_node(self, name, heuristic=0):
        if name not in self.nodes:
            self.nodes[name] = Node(name, heuristic)
        return self.nodes[name]

    def add_edge(self, from_node_name, to_node_name, weight, bidirectional=False):
        if from_node_name not in self.nodes:
            self.add_node(from_node_name)
        if to_node_name not in self.nodes:
            self.add_node(to_node_name)

        from_node = self.nodes[from_node_name]
        to_node = self.nodes[to_node_name]

        from_node.add_edge(to_node, weight)
        if bidirectional:
            to_node.add_edge(from_node, weight)

    def print_graph(self):
        for name, node in self.nodes.items():
            edges_str = ", ".join([f"{neighbor.name} (weight: {weight})" for neighbor, weight in node.edges.items()])
            print(f"Node {name} [Heuristic: {node.heuristic}]: edges to -> {edges_str}")

    def ucs(self, start_name, goal_name):
        if start_name not in self.nodes or goal_name not in self.nodes:
            return None

        # Priority queue stores tuples of (cumulative_cost, current_node_name, path)
        pq = [(0, start_name, [start_name])]
        visited = set()

        while pq:
            current_cost, current_name, path = heapq.heappop(pq)

            if current_name in visited:
                continue

            visited.add(current_name)
            current_node = self.nodes[current_name]

            if current_name == goal_name:
                return path, current_cost

            for neighbor, weight in current_node.edges.items():
                if neighbor.name not in visited:
                    new_cost = current_cost + weight
                    new_path = path + [neighbor.name]
                    heapq.heappush(pq, (new_cost, neighbor.name, new_path))

        return None

# Example usage
if __name__ == "__main__":
    graph = Graph()

    # Add nodes with heuristic values (e.g., estimated cost to goal 'G')
    graph.add_node("S", heuristic=7)
    graph.add_node("A", heuristic=6)
    graph.add_node("B", heuristic=2)
    graph.add_node("C", heuristic=1)
    graph.add_node("G", heuristic=0)

    # Add weighted edges between nodes
    graph.add_edge("S", "A", weight=1, bidirectional=True)
    graph.add_edge("S", "B", weight=4, bidirectional=True)
    graph.add_edge("A", "B", weight=2, bidirectional=True)
    graph.add_edge("A", "C", weight=5, bidirectional=True)
    graph.add_edge("A", "G", weight=12, bidirectional=True)
    graph.add_edge("B", "C", weight=2, bidirectional=True)
    graph.add_edge("C", "G", weight=3, bidirectional=True)

    # Display the graph
    graph.print_graph()

    # Run Uniform Cost Search
    print("\n--- Running Uniform Cost Search (UCS) ---")
    result = graph.ucs("S", "G")
    if result:
        path, cost = result
        print(f"Path found: {' -> '.join(path)}")
        print(f"Total cost: {cost}")
    else:
        print("No path found.")
```
